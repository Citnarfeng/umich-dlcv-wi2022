## Reshaping, reductions, broadcasting, and matrix multiplication

```python
print(core.sum_positive_entries(torch.tensor([-2, 0, 3, 5])))
print(core.reshape_practice(torch.arange(24)))
print(core.zero_row_min(torch.tensor([[10, 20, 30], [2, 5, 1]])))
print(core.normalize_columns(torch.tensor([[1., 2.], [3., 5.], [4., 9.]])))
x, y = torch.randn(2, 3, 4), torch.randn(2, 4, 5)
loop = core.batched_matrix_multiply(x, y, use_loop=True)
vectorized = core.batched_matrix_multiply(x, y, use_loop=False)
print('Batched outputs:', loop.shape, vectorized.shape)
a, b = torch.randn(8, 4), torch.randn(4, 6)
print('CPU product:', core.mm_on_cpu(a, b))
if torch.cuda.is_available():
    print('GPU product returned to CPU:', core.mm_on_gpu(a, b))
```
